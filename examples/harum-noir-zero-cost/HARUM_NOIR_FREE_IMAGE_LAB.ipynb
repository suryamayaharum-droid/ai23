## Texto → imagem canônica de atmosfera


```python
import torch
from diffusers import Flux2KleinPipeline
MODEL='black-forest-labs/FLUX.2-klein-4B'
pipe=Flux2KleinPipeline.from_pretrained(MODEL,torch_dtype=torch.bfloat16)
pipe.enable_model_cpu_offload()
prompt='''Intimate nocturnal charcoal artist atelier, carbon black and graphite, heavyweight aged ivory paper, muted aged-gold accent, mirror, charcoal dust, books, taped anatomy studies, quiet documentary realism, lateral chiaroscuro, 50mm natural perspective, much negative space, no text, no logos, no bright colors, no corporate aesthetic.'''
image=pipe(prompt=prompt,num_inference_steps=4,guidance_scale=1.0,height=1024,width=576,generator=torch.Generator('cpu').manual_seed(2317)).images[0]
image.save('/content/HARUM_NOIR_ATMOSPHERE_01.png')
display(image)
```

## Edição com FACE LOCK
Envie a referência canônica e peça mudança de cenário/roupa sem redesenhar a pessoa do zero.


```python
from google.colab import files
from diffusers.utils import load_image
u=files.upload(); ref_path='/content/'+next(iter(u.keys())); ref=load_image(ref_path)
edit_prompt='''Keep exactly the same adult woman, same face, same dark eyes, same black chin-length bob and natural facial proportions. Place her in a quiet nocturnal charcoal atelier. Matte black clothing, one tiny aged-gold detail, graphite dust and ivory paper. Documentary realism, subtle lateral light. Do not change identity, face shape, hair length, age, ethnicity, makeup or body proportions.'''
edited=pipe(prompt=edit_prompt,image=[ref],num_inference_steps=4,guidance_scale=1.0,height=1024,width=576,generator=torch.Generator('cpu').manual_seed(2317)).images[0]
edited.save('/content/HARUM_NOIR_FACELOCK_SCENE_01.png')
display(edited)
```
