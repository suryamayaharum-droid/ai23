```python
from google.colab import files
uploaded=files.upload()
IMAGE=next(iter(uploaded.keys()))
```

```python
DURATION=8; FPS=30; OUT='HARUM_NOIR_SHORT.mp4'; frames=int(DURATION*FPS)
vf=('scale=1080:1920:force_original_aspect_ratio=increase,crop=1080:1920,'+f"zoompan=z='min(zoom+0.00045,1.055)':x='iw/2-(iw/zoom/2)+sin(on/22)*3':y='ih/2-(ih/zoom/2)+cos(on/29)*2':d={frames}:s=1080x1920:fps={FPS},"+'eq=contrast=1.035:brightness=-0.02:saturation=0.82,vignette=PI/5,noise=alls=2.2:allf=t,'+f'fade=t=in:st=0:d=0.35,fade=t=out:st={DURATION-0.45}:d=0.45')
import subprocess
subprocess.run(['ffmpeg','-y','-loop','1','-i',IMAGE,'-vf',vf,'-t',str(DURATION),'-r',str(FPS),'-c:v','libx264','-preset','medium','-crf','18','-pix_fmt','yuv420p','-an',OUT],check=True)
files.download(OUT)
```
